```python
import dash
from dash import dcc, html
import plotly.graph_objects as go
from dash_bootstrap_components.themes import BOOTSTRAP
from hk_chart.hk_daily_interbank_liquidity import hkfigs_generation
from dash.dependencies import Input, Output, State
from datetime import datetime
```

```python
figs= hkfigs_generation()
```

```python
app = dash.Dash(__name__, external_stylesheets=[BOOTSTRAP])
```

```python
layout1 = html.Div([
    html.H1("HK Chart", style={'textAlign': 'center'}),  
    html.Div([
        html.Div([
            html.H2(title, style={'textAlign': 'center'}),
            dcc.Graph(figure=fig)
        ])
        for fig, title in figs
    ])
])
```

```python
layout2 = html.Div([
    html.H1("HK Chart", style={'textAlign': 'center'}),  
    html.Button("Update Charts", id='update-button', n_clicks=0),
    html.Div(id='update-time', children="Last update time: ", style={'textAlign': 'left', 'margin': '10px'}),
    html.Div(id='graphs-container',children=[
        html.Div([
            html.H2(title, style={'textAlign': 'center'}),
            dcc.Graph(figure=fig)
        ])
        for fig, title in figs
    ])
])

@app.callback(
        [Output('graphs-container', 'children'), Output('update-time', 'children')],
        [Input('update-button', 'n_clicks')],
        [State('graphs-container', 'children')]
)
def update_graphs(n_clicks, current_children):
        if n_clicks > 0:
            try:
                figs = hkfigs_generation()
                return [
                    html.Div([
                        html.H2(title, style={'textAlign': 'center'}),
                        dcc.Graph(figure=fig)
                    ])
                    for fig, title in figs
                ], f"last update time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}"
            except Exception as e:
                print(f"Update failed: {str(e)}")
                return current_children, f"last update time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}"
        return current_children, f"last update time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}"
    
```

```python
layout3 = html.Div([
    html.H1("HK Chart", style={'textAlign': 'center'}),  
    html.Div(f"Last update time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}", style={'textAlign': 'left', 'margin': '10px'}),
    html.Div([
        html.Div([
            html.H2(title, style={'textAlign': 'center'}),
            dcc.Graph( figure=fig)
        ])
        for fig, title in hkfigs_generation()
    ])
])
```

```python
app.layout = layout3
```

```python
app.run(debug=True, host='127.0.0.1', port=8050)
```
